```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import os
```

```python
# Project directory
PROJECT_DIR = os.path.abspath("..")

# Dataset path
RAW_DATA_PATH = os.path.join(
    PROJECT_DIR,
    "data",
    "raw",
    "ChargingRecords.csv"
)

# Cleaned dataset path
CLEANED_DATA_PATH = os.path.join(
    PROJECT_DIR,
    "data",
    "cleaned",
    "ChargingRecords_Cleaned.csv"
)

print("Project Directory:", PROJECT_DIR)

print("Raw Dataset Path:", RAW_DATA_PATH)

print("Cleaned Dataset Path:", CLEANED_DATA_PATH)
```

```python
df = pd.read_csv(RAW_DATA_PATH)

df.head()
```

```python
print("Number of Rows:", df.shape[0])

print("Number of Columns:", df.shape[1])
```

```python
numerical_columns = df.select_dtypes(
    include=["int64", "float64"]
).columns

print("Numerical Columns:")

print(numerical_columns)
```

```python
categorical_columns = df.select_dtypes(
    include=["object"]
).columns

print("Categorical Columns:")

print(categorical_columns)
```

```python
# Analyze categorical features
for column in categorical_columns:

    print("\nColumn:", column)

    print("Unique Values:", df[column].nunique())

    print(df[column].value_counts().head(10))
```

```python
datetime_columns = [
    "StartDatetime",
    "EndDatetime"
]

for column in datetime_columns:

    df[column] = pd.to_datetime(
        df[column],
        errors="coerce"
    )
```

```python
df[datetime_columns].dtypes
```

```python
# charging duration
df["CalculatedDuration"] = (
    df["EndDatetime"] - df["StartDatetime"]
).dt.total_seconds() / 60

df[[
    "StartDatetime",
    "EndDatetime",
    "Duration",
    "CalculatedDuration"
]].head()
```

```python
# data quality validation
print(
    "Negative Duration:",
    (df["Duration"] < 0).sum()
)

print(
    "Negative Demand:",
    (df["Demand"] < 0).sum()
)

print(
    "End Before Start:",
    (df["EndDatetime"] < df["StartDatetime"]).sum()
)

print(
    "Invalid Start Datetime:",
    df["StartDatetime"].isnull().sum()
)

print(
    "Invalid End Datetime:",
    df["EndDatetime"].isnull().sum()
)
```

```python
print("Before Removing Duplicates:", df.shape)

df = df.drop_duplicates()

print("After Removing Duplicates:", df.shape)
```

```python
print("Dataset Before Cleaning:", df.shape)

# Remove records with negative duration
df = df[df["Duration"] >= 0]

# Remove records where end time is before start time
df = df[
    df["EndDatetime"] >= df["StartDatetime"]
]

print("Dataset After Cleaning:", df.shape)
```

```python
os.makedirs(
    os.path.dirname(CLEANED_DATA_PATH),
    exist_ok=True
)

df.to_csv(
    CLEANED_DATA_PATH,
    index=False
)

print("Cleaned Dataset Saved Successfully!")
```
